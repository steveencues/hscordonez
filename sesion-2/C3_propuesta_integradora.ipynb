## 3 · Modelos candidatos y validación

**Guía:** Elige **dos o tres** modelos de las dos sesiones y di por qué encajan con estos datos. Describe la **validación** (K-fold, por grupos si hay varias filas por cliente o máquina, temporal si importa el orden, **anidada** si ajustas hiperparámetros) y la **métrica** según el costo del error (¿qué es peor: un falso positivo o un falso negativo?).

### Respuesta

Seleccionaría tres modelos:
Modelo	Justificación
Regresión logística	Modelo inicial sencillo e interpretable para clasificación binaria.
Random Forest	Permite identificar relaciones no lineales y combinar diferentes variables de los sensores.
Isolation Forest	Permite detectar comportamientos anormales sin necesitar una gran cantidad de fallos previamente etiquetados.

Validación:

Utilizaría validación temporal (Time Series Split) para entrenar con mediciones anteriores y evaluar con mediciones posteriores, evitando utilizar información futura durante el entrenamiento.

También consideraría separar los datos por máquina para comprobar que el modelo pueda generalizar a equipos diferentes.

Para ajustar los hiperparámetros utilizaría validación anidada (Nested Cross-Validation) cuando el volumen de datos lo permita.

Métricas:
- Recall: identificar la mayor cantidad posible de averías reales.
- Precision: evitar generar demasiadas alertas falsas.
- F1-score: equilibrar precision y recall.
- PR-AUC: evaluar el rendimiento cuando existen muchas más observaciones normales que fallos.
El error más costoso sería el falso negativo, porque significaría no detectar una avería que podría detener la producción y generar pérdidas económicas.
